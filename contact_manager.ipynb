```python

# Contact Manager
# Dictionary Version


# Displaying the menu options
def show_menu():
    print("\nContact Manager")
    print("1. Add Contact")
    print("2. View Contacts")
    print("3. Search Contact")
    print("4. Update Contact")
    print("5. Delete Contact")
    print("6. Exit")


#Functiionalities for the Contact Manager
contacts = {}

# Adding Contacts to a dictionary
def add_contact(name, phone, email):
    contacts[name] = {'phone': phone, 'email': email}
    print(f"Contact added: {name}")


# Viewing Contacts in the dictionary
def view_contacts():
    if not contacts:
        print('No contacts found.')
    else:
        for name, info in contacts.items():
            print(f"Name: {name}, Phone: {info['phone']}, Email: {info['email']}")


# Searching for a contact in the dictionary
def search_contact(name):
    if name in contacts:
        info = contacts[name]
        print(f"Contact found: Name: {name}, Phone: {info['phone']}, Email: {info['email']}")
    else:
        print(f"Contact '{name}' not found.")


# Updating a contact in the dictionary
def update_contact(name, phone=None, email=None):
    if name in contacts:
        if phone:
            contacts[name]['phone'] = phone
        if email:
            contacts[name]['email'] = email
        print(f'Contact updated: {name}')
    else:
        print(f"Contact '{name}' not found.")

# Deleting a contact from the dictionary
def delete_contact(name):
    if name in contacts:
        del contacts[name]
        print(f'Contact deleted: {name}')
    else:
        print(f"Contact '{name}' not found.")


# Main loop to run the Contact Manager
if __name__ == "__main__":
    while True:
        show_menu()
        choice = input("Please enter your choice(1-6): ")
        if choice == '1':
            name = input("Enter name: ")
            name = name.strip().lower()
            phone = input("Enter phone: ")
            email = input("Enter email: ")
            add_contact(name, phone, email)
        elif choice == '2':
            view_contacts()
        elif choice == '3':
            name = input("Enter the name you want to search: ")
            name = name.strip().lower()
            
            search_contact(name)
        elif choice == '4':
            name = input("Enter the name you want to update: ")
            name = name.strip().lower()
            new_phone = input("Enter new phone (leave blank to keep current): ")
            new_email = input("Enter new email (leave blank to keep current): ")
            update_contact(name, new_phone, new_email)
        elif choice == '5':
            name = input("Enter the name you want to delete: ")
            name = name.strip().lower()
            delete_contact(name)
        elif choice == '6':
            print("Exiting Contact Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

```

```python

# Contact Manager
# List Version - with duplicated names detected


# Displaying the menu options
def show_menu():
    print("\nContact Manager")
    print("1. Add Contact")
    print("2. View Contacts")
    print("3. Search Contact")
    print("4. Update Contact")
    print("5. Delete Contact")
    print("6. Exit")


# Functionalities for the Contact Manager
contacts = []

# Adding Contacts to a list of dictionaries
def add_contact(name, phone, email):
    contacts.append({'name': name, 'phone': phone, 'email': email})
    print(f"Contact added: {name}")


# Viewing Contacts in the list
def view_contacts():
    if not contacts:
        print('No contacts found.')
    else:
        for contact in contacts:
            print(f"Name: {contact['name']}, Phone: {contact['phone']}, Email: {contact['email']}")


# Searching for a contact in the list
def search_contact(name):
    found = False

    for contact in contacts:
        if contact['name'] == name:
            print(
                f"Name: {contact['name']}, "
                f"Phone: {contact['phone']}, "
                f"Email: {contact['email']}"
            )
            found = True

    if not found:
        print(f"Contact '{name}' not found.")


# Updating a contact in the list
def update_contact(name, phone=None, email=None):
    matches = []

    for contact in contacts:
        if contact['name'] == name:
            matches.append(contact)

    if len(matches) == 0:
        print(f"Contact '{name}' not found.")

    elif len(matches) == 1:
        # update the only match
        match = matches[0]
        if phone:
            match['phone'] = phone
        if email:
            match['email'] = email
        print(f'Contact updated: {name}')
    else:
        # show all matches and ask user to choose
        for index, contact in enumerate(matches, start=1):
            print(
            f"{index}. "
            f"Name: {contact['name']}, "
            f"Phone: {contact['phone']}, "
            f"Email: {contact['email']}"
            )
        choice = int(input("Which contact do you want to update? "))
        if 1 <= choice <= len(matches):
            match = matches[choice - 1]
            if phone:
                match['phone'] = phone
            if email:
                match['email'] = email
            print(f'Contact updated: {name}')
        else:
            print("Invalid choice.")
            return

    # Deleting a contact from the list
    def delete_contact(name):
        matches = []

        for contact in contacts:
            if contact['name'] == name:
                matches.append(contact)

        if len(matches) == 0:
            print(f"Contact '{name}' not found.")
        elif len(matches) == 1:
            # delete the only match
            match = matches[0]
            contacts.remove(match)
            print(f'Contact deleted: {name}')
        else:
            # show all matches and ask user to choose
            for index, contact in enumerate(matches, start=1):
                print(
                f"{index}. "
                f"Name: {contact['name']}, "
                f"Phone: {contact['phone']}, "
                f"Email: {contact['email']}"
                )
            choice = int(input("Which contact do you want to delete? "))
            if 1 <= choice <= len(matches):
                match = matches[choice - 1]
                contacts.remove(match)
                print(f'Contact deleted: {name}')
            else:
                print("Invalid choice.")
                return


# Main loop to run the Contact Manager
if __name__ == "__main__":
    while True:
        show_menu()
        choice = input("Please enter your choice(1-6): ")
        if choice == '1':
            name = input("Enter name: ")
            name = name.strip().lower()
            phone = input("Enter phone: ")
            email = input("Enter email: ")
            add_contact(name, phone, email)
        elif choice == '2':
            view_contacts()
        elif choice == '3':
            name = input("Enter the name you want to search: ")
            name = name.strip().lower()
            
            search_contact(name)
        elif choice == '4':
            name = input("Enter the name you want to update: ")
            name = name.strip().lower()
            new_phone = input("Enter new phone (leave blank to keep current): ")
            new_email = input("Enter new email (leave blank to keep current): ")
            update_contact(name, new_phone, new_email)
        elif choice == '5':
            name = input("Enter the name you want to delete: ")
            name = name.strip().lower()
            delete_contact(name)
        elif choice == '6':
            print("Exiting Contact Manager. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")
```
